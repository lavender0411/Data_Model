```python
SEED = 42
NUM_FOLDS = 10

X = train.drop(['quality', 'Id'], axis = 1)
y = train['quality']

testing = test.drop('Id', axis = 1)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

## Model I : Xgboost 

```python
# 參數選擇
def objective_class(trial, train = X, target = y):
    
    le = LabelEncoder()
    target_trans = le.fit_transform(target)
    
    smote = SMOTE(random_state=SEED)

    # 使數據平衡平均分配
    kf = StratifiedKFold(n_splits=NUM_FOLDS, random_state=SEED, shuffle=True)
    
    params = {
        'random_state': SEED,
        'nthread': -1, 
        'objective': 'multi:softmax',
        'num_class': len(np.unique(target_trans)),
        
        'eta': trial.suggest_float('eta', 0.1, 1),
        'gamma': trial.suggest_float('gamma', 0, 5),  
        'max_depth': trial.suggest_int('max_depth', 3, 8),
        'min_child_weight': trial.suggest_int('min_child_weight', 1, 10), 
        'max_delta_step': trial.suggest_float('max_delta_step', 1, 10),
        'subsample': trial.suggest_float('subsample', 0.1, 1), 
        'colsample_bytree': trial.suggest_float('colsample_bytree', 0.1, 1), 
        'scale_pos_weight': trial.suggest_float('scale_pos_weight', 1, 5), 
        'lambda': trial.suggest_float('lambda', 0, 10), 
        'alpha': trial.suggest_float('alpha', 0, 10), 
     }
    
    scores = []
    
    for f, (train_idx, val_idx) in enumerate(kf.split(train, target)):
        
        train_df, val_df = train.iloc[train_idx], train.iloc[val_idx]
        train_target, val_target = target_trans[train_idx], target_trans[val_idx]
        
        X_sam, y_sam = smote.fit_resample(train_df, train_target)

        trn_data = xgb.DMatrix(X_sam, y_sam)
        val_data = xgb.DMatrix(val_df, val_target)



        model = xgb.train(params, trn_data, num_boost_round=300,
                          evals=[(val_data, 'val_data')], verbose_eval=False, early_stopping_rounds=20)

        y_pred = model.predict(val_data)
        
        scores.append(f1_score(val_target, y_pred, average='weighted'))

    return np.mean(scores)

study = optuna.create_study(direction='maximize')
study.optimize(objective_class, n_trials = 20)

print('Best trial parameters:', study.best_params)
print('Best score:', study.best_value)
```

```python
optuna.visualization.plot_optimization_history(study)
```

```python
from optuna.visualization import plot_param_importances
plot_param_importances(study)
```

```python
skf = StratifiedKFold(n_splits=NUM_FOLDS, shuffle=True, random_state=42)

accuracy_list = []
f1_list = []
conf_matrices = []
fold_reports = []

for fold, (train_idx, val_idx) in enumerate(skf.split(X, y), 1):
    X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
    y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]
    
    smote = SMOTE(random_state=SEED)
    X_train_res, y_train_res = smote.fit_resample(X_train, y_train)
    
    params = {
        'random_state': SEED,
        'nthread': -1, 
        'objective': 'multi:softmax',
        'num_class': y.max()+1,
        'eta': 0.11135814323884936, 
        'gamma': 1.023353180595603, 
        'max_depth': 8, 
        'min_child_weight': 4, 
        'max_delta_step': 7.268723373754219, 
        'subsample': 0.8071594355175286, 
        'colsample_bytree': 0.5580538052779088, 
        'scale_pos_weight': 4.100240837640738, 
        'lambda': 4.335566616207682, 
        'alpha': 9.897154115780449
    }

    trn_data = xgb.DMatrix(X_train_res, label=y_train_res)
    val_data = xgb.DMatrix(X_val, y_val)
    test_data = xgb.DMatrix(X_val)

    model = xgb.train(params, trn_data, num_boost_round=300,
                          evals=[(val_data, 'val_data')], verbose_eval=False, early_stopping_rounds=20)
    
    y_pred = model.predict(test_data)
    
    report_dict = classification_report(y_val, y_pred, output_dict=True)
    fold_reports.append(report_dict)
    
    acc = accuracy_score(y_val, y_pred)
    f1 = f1_score(y_val, y_pred, average='weighted')
    
    accuracy_list.append(acc)
    f1_list.append(f1)
    conf_matrices.append(confusion_matrix(y_val, y_pred))
```

```python
all_classes = sorted(list(set(y)))
metrics = ['precision', 'recall', 'f1-score']

avg_scores = {cls: {m:0 for m in metrics} for cls in all_classes}

for fold_report in fold_reports:
    for cls in map(str, all_classes):
        for m in metrics:
            avg_scores[int(cls)][m] += fold_report[cls][m] / NUM_FOLDS

avg_df = pd.DataFrame(avg_scores).T
avg_df = avg_df.reset_index().rename(columns={'index':'Quality'}).round(3)
avg_df
```

```python
report_long = avg_df.melt(id_vars='Quality', var_name='Metric', value_name='Score')

plt.figure(figsize=(10,6))
sns.barplot(x='Quality', y='Score', hue='Metric', data=report_long)
plt.title('Average Precision / Recall / F1-score by Quality (StratifiedKFold)')
plt.ylim(0,1)
plt.xlabel('Quality')
plt.ylabel('Score')
plt.legend(loc='lower right')
plt.show()
```

```python

print(f"\nMean Accuracy: {np.mean(accuracy_list):.3f}")
print(f"Mean Weighted F1: {np.mean(f1_list):.3f}")

total_cm = sum(conf_matrices)

plt.figure(figsize=(10,8))
sns.heatmap(total_cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=np.unique(y),
            yticklabels=np.unique(y))
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Total Confusion Matrix (5-Fold StratifiedKFold)")
plt.show()
```

```python
# 模型預測
def model_predict_class(train = X, target = y, test = testing):
    
    # 使數據平衡平均分配
    kf = StratifiedKFold(n_splits=NUM_FOLDS, random_state=SEED, shuffle=True)
    
    params = {
        'random_state': SEED,
        'nthread': -1, 
        'objective': 'multi:softmax',
        'num_class': y.max()+1,
        'eta': 0.11135814323884936, 
        'gamma': 1.023353180595603, 
        'max_depth': 8, 
        'min_child_weight': 4, 
        'max_delta_step': 7.268723373754219, 
        'subsample': 0.8071594355175286, 
        'colsample_bytree': 0.5580538052779088, 
        'scale_pos_weight': 4.100240837640738, 
        'lambda': 4.335566616207682, 
        'alpha': 9.897154115780449
    }

    trn_data = xgb.DMatrix(train, label=target)
    test_data = xgb.DMatrix(test)

    model = xgb.train(params, trn_data, num_boost_round=300, verbose_eval=False)

    pred_data = model.predict(test_data)
        
    return pred_data

pred_data = model_predict_class()
```

## Model II : Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from imblearn.pipeline import Pipeline

pipeline = Pipeline([
    ('smote', SMOTE(random_state=42)),
    ('rf', RandomForestClassifier(random_state=42, n_jobs=-1))
])


from sklearn.model_selection import GridSearchCV

param_grid = {
    'smote__k_neighbors': [3, 5, 7],
    'rf__n_estimators': [200, 400],
    'rf__max_depth': [None, 10, 20],
    'rf__min_samples_split': [2, 5],
    'rf__min_samples_leaf': [1, 2]
}

grid = GridSearchCV(
    pipeline,
    param_grid=param_grid,
    scoring='f1_weighted',
    cv=3,
    n_jobs=-1,
    verbose=2
)

grid.fit(X_train, y_train)

print("Best Parameters:", grid.best_params_)
print("Best Score (f1_weighted):", grid.best_score_)
```

```python
best_model = grid.best_estimator_
y_pred = best_model.predict(X_test)

classification_report_df = pd.DataFrame(classification_report(y_test, y_pred, output_dict=True)).transpose()
classification_report_df
```

```python
print("accuracy_score：\n", accuracy_score(y_test, y_pred))
print("confusion_matrix\n", confusion_matrix(y_test, y_pred))
```

```python

# 混淆矩陣圖
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(6,6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')

# 每類 Precision / Recall / F1 條圖
report = pd.DataFrame(classification_report(y_test, y_pred, output_dict=True)).T[:-3]  # 去掉 avg
metrics = ['precision', 'recall', 'f1-score']

plt.figure(figsize=(6,6))
report[metrics].plot(kind='bar')
plt.title('Precision / Recall / F1-score')
plt.xlabel('Quality')
plt.ylabel('Score')
plt.legend(loc='lower right')
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# 特徵重要性圖
importances = best_model.named_steps['rf'].feature_importances_
indices = np.argsort(importances)[::-1]
features = X.columns

plt.figure(figsize=(6,6))
sns.barplot(x=importances[indices], y=features[indices])
plt.title('Feature Importance')
plt.xlabel('Importance')
plt.ylabel('Feature')
plt.show()
```

## Model III : Linear Reg

```python
# Log transform
features_to_transform = ['residual sugar', 'total sulfur dioxide', 'free sulfur dioxide']

X_train_log = X_train.copy()
X_test_log = X_test.copy()

for col in features_to_transform:
    X_train_log[col] = np.log1p(X_train[col])
    X_test_log[col] = np.log1p(X_test[col])
    
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
sns.histplot(X_train['residual sugar'], bins=30, kde=True)
plt.subplot(1, 2, 2)
sns.histplot(X_test_log['residual sugar'], bins=30, kde=True)
```

```python
# StandardScaler
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train_log)
X_test_scaled = scaler.transform(X_test_log)

X_train_scaled_df = pd.DataFrame(X_train_scaled, columns=X.columns)

plt.figure(figsize=(12,4))
plt.subplot(1,2,1)
sns.boxplot(data=X_train)
plt.xticks(rotation=90)
plt.title('Before Scaling')

plt.subplot(1,2,2)
sns.boxplot(data=X_train_scaled_df)
plt.xticks(rotation=90)
plt.title('After StandardScaler')
plt.show()
```

```python
# PCA
from sklearn.decomposition import PCA

pca = PCA(n_components='mle')
X_train_pca = pca.fit_transform(X_train_scaled)
X_test_pca = pca.transform(X_test_scaled)

plt.figure(figsize=(12,4))
plt.subplot(121)
plt.scatter(X_train['alcohol'], X_train['total sulfur dioxide'], c=y_train, cmap='viridis', alpha=0.7)
plt.xlabel('Alcohol')
plt.ylabel('total sulfur dioxide')
plt.title('Before PCA')

plt.subplot(122)
plt.scatter(X_train_pca[:, 0], X_train_pca[:, 1], c=y_train, cmap='viridis', alpha=0.7)
plt.xlabel('Component 1')
plt.ylabel('Component 2')
plt.title('After PCA')

plt.tight_layout()
plt.show()
```

```python
# LinearRegression
from sklearn.linear_model import LinearRegression

reg = LinearRegression().fit(X_train_pca, y_train)
y_pred = reg.predict(X_test_pca)

from sklearn.metrics import mean_squared_error, r2_score
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("MSE:", mse)
print("R²:", r2)
print("Explained Variance Ratio (PCA):", pca.explained_variance_ratio_)
```

## 其他：使用 XGBClassifier

```python
# XGBClassifier
def model_predict_classifier(train = X, target = y, test = testing):

    le = LabelEncoder()

    test_pred = 0
    scores = []


    kf = StratifiedKFold(n_splits=NUM_FOLDS, random_state=SEED, shuffle=True)

    for f, (train_idx, val_idx) in enumerate(kf.split(train, target)):
        target_trans = le.fit_transform(target)
        
        X_train, X_test = train.iloc[train_idx], train.iloc[val_idx]
        y_train, y_test = target_trans[train_idx], target_trans[val_idx]

        xgb_clf = xgb.XGBClassifier(
            n_estimators = 500,
            learning_rate = 0.05,
            eval_metric = "mlogloss")

        xgb_clf.fit(X_train, y_train, verbose = False)

        pred_data = xgb_clf.predict(X_test)

        scores.append(accuracy_score(y_test, le.inverse_transform(pred_data)))

        test_pred += le.inverse_transform(xgb_clf.predict(test))/NUM_FOLDS

    return test_pred, np.mean(scores)

test_pred, scores = model_predict_classifier()

sample_submission['quality'] = test_pred.astype(int)
sample_submission.to_csv('subsci_df_classifier.csv', index=False)
sample_submission
```

```python
def model_predict_reg(train = X, target = y, test = testing):
    
    test_pred = 0
    scores = []

    kf = StratifiedKFold(n_splits=NUM_FOLDS, random_state=SEED, shuffle=True)
    
    params = {
        'random_state': SEED,
        'objective': 'reg:squarederror',
        'nthread': -1, 
        
        'eta': 0.24383670805100566, 
        'gamma': 0.6148174565208251, 
        'max_depth': 5, 
        'min_child_weight': 12, 
        'max_delta_step': 2.1559070673491467, 
        'subsample': 0.8383847654915885, 
        'colsample_bytree': 0.6500674760300006, 
        'scale_pos_weight': 0.8647482462747366, 
        'lambda': 10.499277990812498, 
        'alpha': 24.532310710495764    
    }


    for f, (train_idx, val_idx) in enumerate(kf.split(train, target)):
        
        X_train, X_test = train.iloc[train_idx], train.iloc[val_idx]
        y_train, y_test = target[train_idx], target[val_idx]
        
        trn_data = xgb.DMatrix(X_train, label=y_train)
        val_data = xgb.DMatrix(X_test, label=y_test)
        test_data = xgb.DMatrix(test)

        model = xgb.train(params, trn_data, num_boost_round=100)

        val_pred_data = model.predict(val_data)

        scores.append(mean_squared_error(y_test, val_pred_data.round().astype('int')))
        
        pred_data = model.predict(test_data)

        test_pred += pred_data/NUM_FOLDS

    return test_pred, np.mean(scores)

test_pred, scores = model_predict_reg()

sample_submission['quality'] = test_pred.round().astype(int)
sample_submission.to_csv('subsci_df_reg.csv', index=False)
sample_submission
```
